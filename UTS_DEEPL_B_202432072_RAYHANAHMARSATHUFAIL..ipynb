Algoritma yang dipilih adalah LSTM (Long Short-Term Memory) karena memiliki mekanisme gate (forget, input, output) dan cell state yang mampu menjaga informasi jangka panjang serta mengurangi masalah vanishing gradient yang umum terjadi pada RNN standar, sehingga cocok untuk data berpola musiman.

  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>

```python
#202432072_RayhanahMarsaThufail
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
import copy
np.random.seed(42)
torch.manual_seed(42)
print('PyTorch versi:', torch.__version__)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432072_RayhanahMarsaThufail
df = pd.read_csv('daily-minimum-temperatures-in-me.csv', engine='python', skipfooter=2)
df.columns = ['Date', 'Temp']
df['Date'] = pd.to_datetime(df['Date'])
print(df.shape)
print(df.dtypes)
print(df[df['Temp'].astype(str).str.contains(r'\?')])
```

```python
#202432072_RayhanahMarsaThufail
df['Temp'] = df['Temp'].astype(str).str.replace('?', '-', regex=False)
df['Temp'] = pd.to_numeric(df['Temp'])
df = df.sort_values('Date').reset_index(drop=True)
print(df[df['Temp'] < 0])
print('Missing value:', df['Temp'].isna().sum())
print(df['Temp'].describe())
plt.figure(figsize=(12, 4))
plt.plot(df['Date'], df['Temp'], color='#114F6A', linewidth=0.7)
plt.title('Suhu Minimum Harian Melbourne (1981-1990)')
plt.xlabel('Tanggal')
plt.ylabel('Suhu Minimum (°C)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
#202432072_RayhanahMarsaThufail
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
print('Rentang tanggal:', df['Date'].min().date(), 's.d.', df['Date'].max().date())
print('Tanggal duplikat:', df['Date'].duplicated().sum())
print('Jumlah baris 29 Februari:', ((df['Date'].dt.month == 2) & (df['Date'].dt.day == 29)).sum())
tanggal_lengkap = pd.date_range(df['Date'].min(), df['Date'].max())
print('Tanggal yang tidak ada di data:', list(tanggal_lengkap.difference(df['Date']).strftime('%Y-%m-%d')))
print(df.groupby('Year')['Temp'].agg(['count', 'mean', 'std', 'min', 'max']).round(2))
print(df.groupby('Month')['Temp'].mean().round(2))
print('Skewness:', round(df['Temp'].skew(), 3))
for lag in [1, 7, 30, 365]:
    print('Autokorelasi lag', lag, ':', round(df['Temp'].autocorr(lag=lag), 3))
q1, q3 = df['Temp'].quantile(0.25), df['Temp'].quantile(0.75)
iqr = q3 - q1
outlier = df[(df['Temp'] < q1 - 1.5 * iqr) | (df['Temp'] > q3 + 1.5 * iqr)]
print('Jumlah kandidat outlier (IQR):', len(outlier))
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
axes[0].hist(df['Temp'], bins=40, color='#114F6A', edgecolor='white')
axes[0].set_title('Distribusi Suhu Minimum')
axes[0].set_xlabel('Suhu Minimum (°C)')
axes[0].set_ylabel('Frekuensi')
axes[1].boxplot([df[df['Month'] == m]['Temp'] for m in range(1, 13)], patch_artist=True, boxprops=dict(facecolor='#1FA4B1'))
axes[1].set_title('Suhu Minimum per Bulan')
axes[1].set_xlabel('Bulan')
axes[1].set_ylabel('Suhu Minimum (°C)')
plt.tight_layout()
plt.show()
```

```python
#202432072_RayhanahMarsaThufail
values = df['Temp'].values.astype('float32').reshape(-1, 1)
n = len(values)
train_end = int(n * 0.7)
val_end = int(n * 0.8)
t_min = values[:train_end].min()
t_max = values[:train_end].max()
scaled = (values - t_min) / (t_max - t_min)
window = 30
def create_sequences(data, window):
    X, y = [], []
    for i in range(len(data) - window):
        X.append(data[i:i + window])
        y.append(data[i + window])
    return np.array(X), np.array(y)
X_all, y_all = create_sequences(scaled, window)
target_idx = np.arange(window, n)
train_mask = target_idx < train_end
val_mask = (target_idx >= train_end) & (target_idx < val_end)
test_mask = target_idx >= val_end
X_train = torch.tensor(X_all[train_mask], dtype=torch.float32)
y_train = torch.tensor(y_all[train_mask], dtype=torch.float32)
X_val = torch.tensor(X_all[val_mask], dtype=torch.float32)
y_val = torch.tensor(y_all[val_mask], dtype=torch.float32)
X_test = torch.tensor(X_all[test_mask], dtype=torch.float32)
y_test = torch.tensor(y_all[test_mask], dtype=torch.float32)
dates_test = df['Date'].values[val_end:]
print('Pembagian data (hari): train', train_end, '| val', val_end - train_end, '| test', n - val_end)
print('X_train:', X_train.shape, '| X_val:', X_val.shape, '| X_test:', X_test.shape)
print('y_train:', y_train.shape, '| y_val:', y_val.shape, '| y_test:', y_test.shape)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432072_RayhanahMarsaThufail
class SequenceModel(nn.Module):
    def __init__(self, input_size=1, hidden_size=64, dropout=0.2):
        super(SequenceModel, self).__init__()
        self.lstm = nn.LSTM(input_size, hidden_size, batch_first=True)
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(hidden_size, 1)
    def forward(self, x):
        out, (h_n, c_n) = self.lstm(x)
        out = self.dropout(out[:, -1, :])
        return self.fc(out)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432072_RayhanahMarsaThufail
epochs = 50
learning_rate = 0.005
batch_size = 64
model = SequenceModel(input_size=1, hidden_size=64, dropout=0.2)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)
train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=batch_size, shuffle=True)
train_losses = []
val_losses = []
best_val = float('inf')
best_epoch = 0
best_state = None
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for xb, yb in train_loader:
        optimizer.zero_grad()
        pred = model(xb)
        loss = criterion(pred, yb)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * xb.size(0)
    epoch_loss = running_loss / len(X_train)
    train_losses.append(epoch_loss)
    model.eval()
    with torch.no_grad():
        val_loss = criterion(model(X_val), y_val).item()
    val_losses.append(val_loss)
    if val_loss < best_val:
        best_val = val_loss
        best_epoch = epoch + 1
        best_state = copy.deepcopy(model.state_dict())
    if (epoch + 1) % 10 == 0:
        print(f'Epoch [{epoch+1}/{epochs}] Train Loss: {epoch_loss:.5f} | Val Loss: {val_loss:.5f}')
model.load_state_dict(best_state)
print(f'Model terbaik: epoch {best_epoch} dengan Val Loss {best_val:.5f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432072_RayhanahMarsaThufail
plt.figure(figsize=(10, 4))
plt.plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
plt.plot(val_losses, color='#1FA4B1', linewidth=2, linestyle='--', label='Validation Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

```python
#202432072_RayhanahMarsaThufail
model.eval()
with torch.no_grad():
    pred_scaled = model(X_test).numpy()
pred = pred_scaled * (t_max - t_min) + t_min
actual = y_test.numpy() * (t_max - t_min) + t_min
rmse = np.sqrt(np.mean((pred - actual) ** 2))
mae = np.mean(np.abs(pred - actual))
r2 = 1 - np.sum((actual - pred) ** 2) / np.sum((actual - actual.mean()) ** 2)
baseline_pred = np.concatenate([values[val_end - 1:val_end], actual[:-1]])
naive_rmse = np.sqrt(np.mean((actual - baseline_pred) ** 2))
naive_mae = np.mean(np.abs(actual - baseline_pred))
print(f'LSTM      -> RMSE: {rmse:.3f} °C | MAE: {mae:.3f} °C | R2: {r2:.3f}')
print(f'Baseline  -> RMSE: {naive_rmse:.3f} °C | MAE: {naive_mae:.3f} °C')
plt.figure(figsize=(12, 5))
plt.plot(dates_test, actual, color='#114F6A', linewidth=1.2, label='Data Aktual')
plt.plot(dates_test, pred, color='#1FA4B1', linewidth=1.2, label='Prediksi LSTM')
plt.title('Perbandingan Data Aktual vs Prediksi LSTM pada Data Uji')
plt.xlabel('Tanggal')
plt.ylabel('Suhu Minimum (°C)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
plt.figure(figsize=(12, 4))
plt.plot(dates_test[:100], actual[:100], color='#114F6A', marker='o', markersize=3, label='Data Aktual')
plt.plot(dates_test[:100], pred[:100], color='#1FA4B1', marker='o', markersize=3, label='Prediksi LSTM')
plt.title('Zoom: 100 Hari Pertama Data Uji')
plt.xlabel('Tanggal')
plt.ylabel('Suhu Minimum (°C)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
